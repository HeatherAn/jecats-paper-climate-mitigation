```python
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
altitude = np.repeat(350, 11)
time1    = np.arange(len(altitude))

altitude2 = np.array([350, 350, 350, 330, 330, 330, 330, 330, 350, 350, 350])
time2     = np.arange(len(altitude2))

fig, ax = plt.subplots(figsize=(4, 4))

ax.plot(time1, altitude, label="Filed", color=sns.color_palette("Purples")[3], linewidth=7)
ax.plot(time2, altitude2, label="Contrail-optimised", color=sns.color_palette("Greens")[2], linewidth=7)

ax.fill_between(time1, altitude, altitude2, alpha=0.3, color=sns.color_palette("Greens")[2], label="Deviation")

bg = "#1c1c1c"
ax.axis("off")
fig.patch.set_facecolor(bg)
ax.set_facecolor(bg)

ax.set_ylim([320, 360])

plt.savefig("../reports/logo_darkmode.png", dpi=300, bbox_inches="tight")
```
